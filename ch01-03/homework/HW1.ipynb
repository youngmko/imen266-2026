## Part B3 — Monte-Carlo check of your corrected answer

Model (conditionally independent given status): prevalence 0.004,
sensitivity 0.95, false-positive rate 0.04 per test.

```python
# --- B3 skeleton -------------------------------------------------------------
N = 2_000_000
prev, sens, fpr = 0.004, 0.95, 0.04

sick = rng.random(N) < prev
t1 = np.where(sick, rng.random(N) < sens, rng.random(N) < fpr)
t2 = np.where(sick, rng.random(N) < sens, rng.random(N) < fpr)   # cond. indep.

both = t1 & t2
p_hat = sick[both].mean()
n_pos = both.sum()
se = np.sqrt(p_hat * (1 - p_hat) / n_pos)
print(f"P(cancer | ++) ≈ {p_hat:.4f}   95% CI ± {1.96*se:.4f}   (n_++ = {n_pos:,})")

# TODO: compare with your analytic B2 value (posterior-odds route) here:
analytic_b2 = None   # <- fill in
print("analytic:", analytic_b2)
```

## Part C2 — Dependent false positives (benign condition $B$)

Among the **healthy**: benign condition $B$ with prob 0.02 makes each test
positive with prob 0.9; otherwise with prob $p_0$, calibrated so the marginal
FPR stays 0.04. Tests are conditionally independent given (status, $B$).

```python
# --- C2 skeleton -------------------------------------------------------------
N = 2_000_000
prev, sens = 0.004, 0.95
p_B, p_pos_B = 0.02, 0.90

# TODO C2a: calibrate p0 so that  p_B*p_pos_B + (1-p_B)*p0 = 0.04
p0 = np.nan     # <- replace np.nan with your calibrated value

sick = rng.random(N) < prev
B    = (~sick) & (rng.random(N) < p_B)          # benign condition, healthy only

# per-person probability that ONE test comes out positive
p_pos = np.where(sick, sens, np.where(B, p_pos_B, p0))

t1 = rng.random(N) < p_pos
t2 = rng.random(N) < p_pos                       # cond. indep. given (status, B)

both = t1 & t2
print(f"P(cancer | ++) under dependence ≈ {sick[both].mean():.4f}")
print( "  vs Part-B (cond.-indep.) value:", None)   # <- paste your B result

# Suggested figure: bar chart of the two PPVs side by side.
```

## Part C3 (optional) — When does Binomial ≈ Hypergeometric?

```python
# --- C3 skeleton -------------------------------------------------------------
n = 10
Ns = np.array([20, 30, 50, 100, 200, 500, 1000, 2000, 5000])

def tv_distance(N):
    K = int(np.ceil(0.15 * N))                     # defectives in the lot
    ks = np.arange(0, n + 1)
    ph = stats.hypergeom.pmf(ks, N, K, n)
    pb = stats.binom.pmf(ks, n, 0.15)
    return 0.5 * np.abs(ph - pb).sum()

# TODO: compute tv for each N, plot on log-x, and state your "safe N" rule.
```
